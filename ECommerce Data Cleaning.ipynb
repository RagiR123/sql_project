```python
product_details = pd.read_csv(r'C:\Users\User\Desktop\P\ECommerceSales\product_details.csv')
```

```python
product_details.head()
```

```python
product_details.shape
```

```python
product_details.isnull().sum()
```

```python
product_details= product_details.dropna(axis=1,how='all')
```

```python
product_details.isnull().sum()
```

```python
product_details.columns = (product_details.columns.str.strip().str.lower().str.replace(" ","_"))
```

```python
product_details.info()
```

```python
product_details['weight_value'] = pd.to_numeric(product_details['shipping_weight'].str.extract(r"(\d+(?:\.\d+)?)")[0],errors='coerce')
product_details['weight_unit'] = (product_details['shipping_weight'].str.extract(r"([a-zA-Z]+)")[0].str.lower())

product_details.head()
```

```python
product_details['weight_lb'] = product_details['weight_value']
product_details.loc[product_details['weight_unit'].str.contains('ounce',na=False),'weight_lb'] = (
product_details.loc[product_details['weight_unit'].str.contains('ounce',na=False),'weight_value']/16)
```

```python
product_details.head()
```

```python
product_details = product_details.drop(columns=['shipping_weight','about_product','product_specification','technical_details','product_dimensions','image','variants','product_url'])
```

```python
product_details.info()
```

```python

product_details['selling_price'] = (product_details['selling_price'].str.replace('$',"",regex=False)
                                  .str.replace(",","",regex=False))
product_details['selling_price'] = pd.to_numeric(product_details['selling_price'],errors='coerce')
product_details['selling_price'] = product_details['selling_price'].fillna(0)
```

```python
product_details
```

```python
product_details.info()
```

```python
long_names = product_details[product_details['product_name'].astype(str).str.len() > 240]
print(len(long_names))
```

```python
product_details[product_details.select_dtypes(include='object').columns] = product_details.select_dtypes(include='object').fillna("Unknown")

product_details[product_details.select_dtypes(exclude='object').columns] = product_details.select_dtypes(exclude='object').fillna(0)

product_details.head()
```

```python
product_details.head()
```

```python
import unicodedata
def clean_ascii(value):
    if pd.isna(value):
        return value
    
    value = str(value)
    #value = unicodedata.normalize("NFKC",value)
    value = value.encode("ascii",errors = "ignore").decode("ascii")
    
    return value.strip()
```

```python
for column in product_details.columns:
    product_details[column] = product_details[column].apply(clean_ascii)
    
print("Total rows :", len(product_details))
print("Total columns :",len(product_details.columns))
```

```python
product_details.to_csv(r'C:\Users\User\Desktop\P\ECommerceSales\cleaned_product_details.csv',index=False,encoding='utf-8-sig')
```
